```python
import pandas as pd
import numpy as np
import psycopg2

# Step 1 - Connect to database
conn = psycopg2.connect(
    host="localhost",
    database="health_db",
    user="postgres",
    password="admin123",
    port="5432"
)
print("✅ Step 1 - Connected to database!")

# Step 2 - Load CSV file
df = pd.read_csv(r"C:\Users\hanna\Downloads\diabetes+130-us+hospitals+for+years+1999-2008\diabetic_data.csv")
df.replace("?", np.nan, inplace=True)
print(f"✅ Step 2 - Data loaded! {len(df)} rows, {len(df.columns)} columns")

# Step 3 - SQL Analysis
print("\n--- ANALYSIS 1: Readmission counts ---")
df1 = pd.read_sql("""
    SELECT readmitted, COUNT(*) as total_patients
    FROM patient_visits
    GROUP BY readmitted
    ORDER BY total_patients DESC
""", conn)
print(df1)

print("\n--- ANALYSIS 2: Age group readmissions ---")
df2 = pd.read_sql("""
    SELECT age, COUNT(*) as total,
    SUM(CASE WHEN readmitted = '<30' THEN 1 ELSE 0 END) as readmitted_30days
    FROM patient_visits
    GROUP BY age
    ORDER BY readmitted_30days DESC
""", conn)
print(df2)

print("\n--- ANALYSIS 3: Average hospital stay by age ---")
df3 = pd.read_sql("""
    SELECT age,
    ROUND(AVG(time_in_hospital)::numeric, 2) as avg_days
    FROM patient_visits
    GROUP BY age
    ORDER BY avg_days DESC
""", conn)
print(df3)

print("\n🎉 ALL ANALYSIS COMPLETE!")
conn.close()
```
