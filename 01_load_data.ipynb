```python
import psycopg2
from psycopg2.extensions import ISOLATION_LEVEL_AUTOCOMMIT

# Connect to default postgres database
conn = psycopg2.connect(
    host="localhost",
    database="postgres", 
    user="postgres",
    password="admin123",
    port="5432"
)
conn.set_isolation_level(ISOLATION_LEVEL_AUTOCOMMIT)
cursor = conn.cursor()

# Create health_db
cursor.execute("CREATE DATABASE health_db")
print("Database health_db created successfully!")

cursor.close()
conn.close()
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql://postgres:admin123@localhost:5432/health_db"
)

df.to_sql("patient_visits", engine, 
          if_exists="replace", index=False)

print(f"SUCCESS! {len(df)} patient records loaded!")
```

```python
# Verify data is in database
import pandas as pd
df_check = pd.read_sql("SELECT COUNT(*) as total_patients FROM patient_visits", engine)
print(df_check)

# Preview first 5 rows from database
df_preview = pd.read_sql("SELECT * FROM patient_visits LIMIT 5", engine)
df_preview
```
